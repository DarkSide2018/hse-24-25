```python
def calculate_degrees(edges):
    degree_count = {}
    
    for v1, v2 in edges:
        if v1 not in degree_count:
            degree_count[v1] = 0
        if v2 not in degree_count:
            degree_count[v2] = 0
        
        degree_count[v1] += 1
        degree_count[v2] += 1
    
    return degree_count

# Пример рёбер графа
edges = [(1, 2),
         (1, 4),
         (2, 3),
         (2,1),
         (2,4),
         (3,2),
         (3,6),
         (3,5),
         (4, 1),
         (4, 2),
         (5,3),
         (6,3)
         ]
degrees = calculate_degrees(edges)
print(degrees)
```

```python
import networkx as nx

# Создаем граф из рёбер
edges = [(1, 2), (1, 4), (2, 3), (2, 1), (2, 4), (3, 2), (3, 6), (3, 5), (4, 1), (4, 2), (5, 3), (6, 3)]
G = nx.Graph(edges)

# Функция для вычисления сходства Жаккара
def jaccard_similarity(G, node1, node2):
    neighbors1 = set(G.neighbors(node1))
    print("neighbors1  = ", neighbors1)
    neighbors2 = set(G.neighbors(node2))
    print("neighbors2  = ", neighbors2)
    intersection = len(neighbors1.intersection(neighbors2))
    print("intersection_length = ", intersection)
    union_dict=neighbors1.union(neighbors2)
    print("union_dict", union_dict)
    union = len(union_dict)
    print("union_length = ", union)
    if union == 0:
        return 0  # Избегаем деления на ноль
    return intersection / union

# Вычисляем сходство Жаккара для заданных пар
pairs = [(1, 3), (5, 6), (1, 4)]
similarities = {pair: jaccard_similarity(G, pair[0], pair[1]) for pair in pairs}

print("Сходство Жаккара для пар вершин:", similarities)
```

```python

import math

# Функция для вычисления индекса Адамика-Адара
def adamic_adar_index(G, node1, node2):
    neighbors1 = set(G.neighbors(node1))
    neighbors2 = set(G.neighbors(node2))
    
    common_neighbors = neighbors1.intersection(neighbors2)
    index = 0.0
    
    for w in common_neighbors:
        degree_w = G.degree(w)
        print("degree_w", degree_w)
        if degree_w > 1:  # Избегаем логарифма от 1
            index += 1 / math.log(degree_w)
    
    return index

# Вычисляем индекс Адамика-Адара для заданных пар
pairs = [(1, 3), (5, 6), (1, 4)]
adamic_adar_indices = {pair: adamic_adar_index(G, pair[0], pair[1]) for pair in pairs}

print("Индекс Адамика-Адара для пар вершин:", adamic_adar_indices)
```

```python

# Функция для вычисления вектора степени графлетов размера 3
def graphlet_degree_vector(G, node):
    neighbors = list(G.neighbors(node))
    print("neighbors: ", neighbors)
    degree_vector = [0] * 4  # Вектор для 4 графлетов размера 3

    # Проверяем все комбинации соседей
    for i in range(len(neighbors)):
        for j in range(i + 1, len(neighbors)):
            n1 = neighbors[i]
            n2 = neighbors[j]
            if G.has_edge(n1, n2):
                degree_vector[2] += 1  # Треугольник (G2)
            else:
                degree_vector[1] += 1  # Линейная структура (G1)

    degree_vector[0] = 1 if len(neighbors) == 0 else 0  # Три отдельные вершины (G0)
    degree_vector[3] = len(neighbors) * (len(neighbors) - 1) // 2 - degree_vector[1] - degree_vector[2]  # Вершина с двумя рёбрами (G3)

    return degree_vector

# Вычисляем вектор степени графлетов для вершин 2, 4 и 5
nodes = [2, 4, 5]
gdv_results = {node: graphlet_degree_vector(G, node) for node in nodes}

print("Векторы степени графлетов для вершин 2, 4 и 5:", gdv_results)
```

```python
import networkx as nx
from itertools import combinations

# Создаем граф из рёбер
edges = [(1, 2), (1, 4), (2, 3), (2, 1), (2, 4), (3, 2), (3, 6), (3, 5), (4, 1), (4, 2), (5, 3), (6, 3)]
G = nx.Graph(edges)

# Функция для вычисления графлетового ядра размера 3
def graphlet_core(G):
    graphlet_counts = {}
    nodes = list(G.nodes())
    
    for node in nodes:
        neighbors = list(G.neighbors(node))
        counts = [0] * 4  # Вектор для 4 графлетов размера 3

        # Проверяем все комбинации соседей
        for n1, n2 in combinations(neighbors, 2):
            if G.has_edge(n1, n2):
                counts[0] += 1  # Треугольник (G2)
            else:
                counts[1] += 1  # Линейная структура (G1)

        counts[2] = 1 if len(neighbors) == 0 else 0  # Три отдельные вершины (G0)
        counts[3] = len(neighbors) * (len(neighbors) - 1) // 2 - counts[0] - counts[1]  # Вершина с двумя рёбрами (G3)

        graphlet_counts[node] = counts

    return graphlet_counts

# Вычисляем графлетово ядро для всего графа
graphlet_cores = graphlet_core(G)

print("Графлетово ядро для графа:", graphlet_cores)
```

```python
import numpy as np
import networkx as nx

# Создаем граф из рёбер
edges = [(5, 1), (1, 3), (1, 4), (3, 2), (3, 1), (2, 3), (2, 4), (4, 1), (4, 2), (1, 5)]
G = nx.Graph(edges)

# Получаем список вершин
nodes = list(G.nodes())
n = len(nodes)

# Создаем матрицу смежности
adjacency_matrix = np.zeros((n, n))

# Заполняем матрицу смежности
for i, node in enumerate(nodes):
    for neighbor in G.neighbors(node):
        j = nodes.index(neighbor)
        adjacency_matrix[i][j] = 1

# Преобразуем в стохастическую матрицу
stochastic_matrix = np.zeros((n, n))

for i in range(n):
    row_sum = np.sum(adjacency_matrix[i])
    if row_sum > 0:
        stochastic_matrix[i] = adjacency_matrix[i] / row_sum

print("Стохастическая матрица смежности:")
print(stochastic_matrix)

```

```python
r = np.array([1, 0, 0, 0, 0])  # Начинаем с вершины 1
next_state = r @ stochastic_matrix
print("Состояние после одного шага:", next_state)
two_steps_state = next_state @ stochastic_matrix
print("Состояние после двух шагов:", two_steps_state)
```

```python
import networkx as nx

# Создаем граф из рёбер
edges = [(5, 1), (1, 3), (1, 4), (3, 2), (3, 1), (2, 3), (2, 4), (4, 1), (4, 2), (1, 5)]
G = nx.Graph(edges)

# Параметры
d = 0.85  # коэффициент затухания
num_iterations = 100  # количество итераций
tolerance = 1.0e-6  # порог для сходимости

# Инициализация PageRank
N = len(G.nodes())
pagerank = {node: 1.0 / N for node in G.nodes()}

# Итеративный расчет PageRank
for _ in range(num_iterations):
    new_pagerank = {}
    for node in G.nodes():
        rank_sum = 0.0
        for neighbor in G.neighbors(node):
            rank_sum += pagerank[neighbor] / len(list(G.neighbors(neighbor)))
        new_pagerank[node] = (1 - d) / N + d * rank_sum

    # Проверка на сходимость
    if all(abs(new_pagerank[node] - pagerank[node]) < tolerance for node in G.nodes()):
        break

    pagerank = new_pagerank

# Округляем до сотых
pagerank_rounded = {node: round(score, 2) for node, score in pagerank.items()}

# Выводим результаты
print("PageRank Scores (до сотых):")
for node, score in pagerank_rounded.items():
    print(f"Вершина {node}: {score}")
```
